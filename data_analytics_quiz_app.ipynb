```python
import tkinter as tk
from tkinter import messagebox
import random

questions = [

    {
        "question": "What is the output of print(10 // 3) in Python?",
        "options": ["3", "3.33", "4", "1"],
        "answer": "3"
    },
    {
        "question": "Which Python library is commonly used for data manipulation and analysis?",
        "options": ["Pandas", "Tkinter", "Pygame", "Flask"],
        "answer": "Pandas"
    },
    {
        "question": "Which function is used to find the number of items in a Python list?",
        "options": ["count()", "length()", "len()", "size()"],
        "answer": "len()"
    },
    {
        "question": "Which Python data structure stores data in key-value pairs?",
        "options": ["List", "Tuple", "Dictionary", "Set"],
        "answer": "Dictionary"
    },
    {
        "question": "What does df.head() return in Pandas?",
        "options": [
            "Last five rows",
            "First five rows",
            "Column names only",
            "Total number of rows"
        ],
        "answer": "First five rows"
    },

    {
        "question": "Which SQL statement is used to retrieve data from a table?",
        "options": ["GET", "SELECT", "FETCH", "RETRIEVE"],
        "answer": "SELECT"
    },
    {
        "question": "Which SQL clause is used to filter rows?",
        "options": ["ORDER BY", "GROUP BY", "WHERE", "SORT BY"],
        "answer": "WHERE"
    },
    {
        "question": "Which SQL function calculates the average of a column?",
        "options": ["SUM()", "AVG()", "MEAN()", "AVERAGE()"],
        "answer": "AVG()"
    },
    {
        "question": "Which SQL clause is used to group rows with the same values?",
        "options": ["GROUP BY", "ORDER BY", "WHERE", "HAVING"],
        "answer": "GROUP BY"
    },
    {
        "question": "Which key uniquely identifies each record in a table?",
        "options": ["Foreign Key", "Primary Key", "Candidate Key", "Index Key"],
        "answer": "Primary Key"
    },

    {
        "question": "Which Excel function is used to calculate the total of a range?",
        "options": ["TOTAL()", "SUM()", "ADD()", "COUNT()"],
        "answer": "SUM()"
    },
    {
        "question": "Which Excel function counts cells containing numbers?",
        "options": ["COUNT()", "COUNTA()", "COUNTIF()", "SUM()"],
        "answer": "COUNT()"
    },
    {
        "question": "Which Excel feature is commonly used to summarize and analyze large datasets?",
        "options": [
            "WordArt",
            "PivotTable",
            "Page Layout",
            "Text Box"
        ],
        "answer": "PivotTable"
    },
    {
        "question": "Which Excel function is commonly used to find a value from another table based on a matching value?",
        "options": ["VLOOKUP()", "SUM()", "ROUND()", "LEFT()"],
        "answer": "VLOOKUP()"
    },
    {
        "question": "Which function can be used to calculate the average in Excel?",
        "options": ["MEAN()", "AVG()", "AVERAGE()", "MID()"],
        "answer": "AVERAGE()"
    },

    {
        "question": "What is the main purpose of data analytics?",
        "options": [
            "To design websites",
            "To extract insights from data",
            "To create operating systems",
            "To write mobile applications"
        ],
        "answer": "To extract insights from data"
    },
    {
        "question": "Which step usually involves removing duplicate records and correcting missing or incorrect values?",
        "options": [
            "Data Cleaning",
            "Data Visualization",
            "Data Export",
            "Data Encryption"
        ],
        "answer": "Data Cleaning"
    },
    {
        "question": "Which chart is generally suitable for showing trends over time?",
        "options": [
            "Pie Chart",
            "Line Chart",
            "Radar Chart",
            "Doughnut Chart"
        ],
        "answer": "Line Chart"
    },
    {
        "question": "What does KPI stand for in data analytics?",
        "options": [
            "Key Performance Indicator",
            "Known Process Information",
            "Key Programming Interface",
            "Knowledge Performance Index"
        ],
        "answer": "Key Performance Indicator"
    },
    {
        "question": "Which measure represents the middle value of an ordered dataset?",
        "options": ["Mean", "Median", "Mode", "Range"],
        "answer": "Median"
    }
]

random.shuffle(questions)

current_question = 0
score = 0

window = tk.Tk()

window.title("Data Analytics Quiz")
window.geometry("750x600")
window.configure(bg="#1e1e2f")
window.resizable(False, False)

title = tk.Label(
    window,
    text="DATA ANALYTICS QUIZ",
    font=("Arial", 24, "bold"),
    bg="#1e1e2f",
    fg="#00d9ff"
)

title.pack(pady=20)


subtitle = tk.Label(
    window,
    text="Python • MySQL • Excel • Data Analytics",
    font=("Arial", 12, "italic"),
    bg="#1e1e2f",
    fg="#cccccc"
)

subtitle.pack()

question_frame = tk.Frame(
    window,
    bg="#2b2b40",
    padx=30,
    pady=25
)

question_frame.pack(
    padx=40,
    pady=20,
    fill="both"
)

question_number = tk.Label(
    question_frame,
    text="",
    font=("Arial", 12, "bold"),
    bg="#2b2b40",
    fg="#00d9ff"
)

question_number.pack(pady=5)

question_label = tk.Label(
    question_frame,
    text="",
    font=("Arial", 15, "bold"),
    bg="#2b2b40",
    fg="white",
    wraplength=650,
    justify="center"
)

question_label.pack(pady=15)

selected_answer = tk.StringVar()

option_buttons = []

for i in range(4):

    button = tk.Radiobutton(
        question_frame,
        text="",
        variable=selected_answer,
        value="",
        font=("Arial", 12),
        bg="#2b2b40",
        fg="white",
        selectcolor="#3b3b55",
        activebackground="#2b2b40",
        activeforeground="#00d9ff",
        anchor="w",
        padx=20,
        pady=8
    )

    button.pack(
        fill="x",
        pady=3
    )

    option_buttons.append(button)

progress_label = tk.Label(
    window,
    text="",
    font=("Arial", 11),
    bg="#1e1e2f",
    fg="#aaaaaa"
)

progress_label.pack(pady=5)

def next_question():

    global current_question, score
    if selected_answer.get() == "":
        messagebox.showwarning(
            "No Answer",
            "Please select an answer before continuing."
        )
        return
        
    if selected_answer.get() == questions[current_question]["answer"]:
        score += 1

    current_question += 1

    if current_question >= len(questions):

        percentage = (score / len(questions)) * 100

        if percentage >= 80:
            message = "Excellent performance!"
        elif percentage >= 60:
            message = "Good job! Keep practicing."
        else:
            message = "Keep practicing and improve your skills."

        messagebox.showinfo(
            "Quiz Completed",
            f"Quiz Completed!\n\n"
            f"Your Score: {score}/{len(questions)}\n"
            f"Percentage: {percentage:.1f}%\n\n"
            f"{message}"
        )

        window.destroy()

    else:
        show_question()

def show_question():

    selected_answer.set("")

    q = questions[current_question]

    question_number.config(
        text=f"Question {current_question + 1}"
    )

    question_label.config(
        text=q["question"]
    )

    progress_label.config(
        text=f"{current_question + 1} / {len(questions)}"
    )

    for i in range(4):

        option_buttons[i].config(
            text=q["options"][i],
            value=q["options"][i]
        )

next_button = tk.Button(
    window,
    text="NEXT QUESTION →",
    font=("Arial", 12, "bold"),
    bg="#00b894",
    fg="white",
    activebackground="#009e7a",
    activeforeground="white",
    width=22,
    height=2,
    relief="flat",
    cursor="hand2",
    command=next_question
)

next_button.pack(pady=15)

def exit_quiz():

    answer = messagebox.askyesno(
        "Exit Quiz",
        "Are you sure you want to exit the quiz?"
    )

    if answer:
        window.destroy()


exit_button = tk.Button(
    window,
    text="EXIT QUIZ",
    font=("Arial", 10, "bold"),
    bg="#e74c3c",
    fg="white",
    activebackground="#c0392b",
    activeforeground="white",
    width=15,
    relief="flat",
    cursor="hand2",
    command=exit_quiz
)

exit_button.pack()

show_question()

window.mainloop()
```
